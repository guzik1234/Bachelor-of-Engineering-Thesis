```python
import plotly.express as px
import pandas as pd
import yfinance as yf

def fetch_financial_data(company='AMZN'):
    """
    This function fetches stock market quotations using yfinance.
    """
    # Pobieramy dane historyczne dla wybranej spółki
    df = yf.download(company)
    return df

df = fetch_financial_data()
df.head()
df = df.reset_index()
df.head()
```

```python
import plotly.express as px
import pandas as pd

# Przygotowujemy dane jako ramkę DataFrame
data = [
    dict(Task='Projekt A', Start="2019-01-01", Finish="2019-06-01"),
    dict(Task='Projekt B', Start="2019-03-01", Finish="2019-07-01"),
    dict(Task='Projekt C', Start="2019-02-01", Finish="2019-05-01"),
    dict(Task='Projekt D', Start="2019-04-01", Finish="2019-10-01"),
]
df_gantt = pd.DataFrame(data)

# Tworzymy wykres Gantta przy użyciu px.timeline
fig = px.timeline(df_gantt, x_start="Start", x_end="Finish", y="Task")
fig.update_yaxes(autorange="reversed")  # Odwracamy oś Y, aby pierwsze zadanie było na górze
fig.show()
```

```python
import plotly.express as px
import pandas as pd

# Przygotowujemy dane jako ramkę DataFrame
data = [
    dict(Task='Projekt A', Start="2019-01-01", Finish="2019-06-01", Complete=10),
    dict(Task='Projekt B', Start="2019-03-01", Finish="2019-07-01", Complete=50),
    dict(Task='Projekt C', Start="2019-02-01", Finish="2019-05-01", Complete=100),
    dict(Task='Projekt D', Start="2019-04-01", Finish="2019-10-01", Complete=70),
]
df_gantt = pd.DataFrame(data)

# Tworzymy wykres Gantta przy użyciu px.timeline bez błędnych parametrów
fig = px.timeline(
    df_gantt,
    x_start="Start",
    x_end="Finish",
    y="Task",
    color="Complete",
    color_continuous_scale=["#eb4034", "#34eb89"]
)

# Ewentualna konfiguracja paska kolorów (colorbar) po utworzeniu wykresu
fig.update_layout(coloraxis_showscale=True)

fig.update_yaxes(autorange="reversed")  # Odwracamy oś Y, aby pierwsze zadanie było na górze
fig.show()
```

```python
l = raw.split('\n')
l = [row.strip().split('/') for row in l][:-1]
code = []
country = []
for item in l:
  code.append(item[0].strip())
  country.append(item[1].strip())
print(code)
print(country)
df = pd.DataFrame({'code': code, 'country': country})
df = df.reset_index()
df.head()
```
